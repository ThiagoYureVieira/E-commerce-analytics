```python
import pandas as pd
import numpy as np

#Carregando csv de avaliações
df_reviews = pd.read_csv("../data/staging/order_reviews.csv", encoding="UTF-8")

display(df_reviews)
```

```python
#Análise Inicial dos dados
print("Quantidade de colunas - ")

display(df_reviews.shape)
display(list(df_reviews.columns))
columns = list(df_reviews.columns)

print("Descrição das tabelas - ")

display(df_reviews.info())
display(df_reviews.describe())
print("Soma dos valores NULL de cada coluna - ")
display(df_reviews.isnull().sum())
print(f"Soma dos valores duplicados: {df_reviews.duplicated(subset=['order_id', 'review_id']).sum()}")
```

```python
#Limpeza dos dados

df_reviews.fillna({'review_comment_title': 'Sem título', 'review_comment_message': 'Sem mensagem'}, inplace=True)

df_reviews['review_comment_title'] = df_reviews['review_comment_title'].str.strip()
df_reviews['review_comment_message'] = df_reviews['review_comment_message'].str.strip()
df_reviews['review_comment_title'] = df_reviews['review_comment_title'].str.capitalize()
df_reviews['review_comment_message'] = df_reviews['review_comment_message'].str.capitalize()
display(df_reviews)
```

```python
df_reviews.to_csv("../data/processed/order_reviews_processed.csv")
```
