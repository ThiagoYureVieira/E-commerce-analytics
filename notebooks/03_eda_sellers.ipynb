```python
import pandas as pd
import numpy as np

#Carregando csv de vendedores
df_sellers = pd.read_csv("../data/staging/sellers.csv", encoding="UTF-8")

display(df_sellers)
```

```python
#Análise Inicial dos dados
print("Quantidade de colunas - ")

display(df_sellers.shape)
display(list(df_sellers.columns))
columns = list(df_sellers.columns)

print("Descrição das tabelas - ")

display(df_sellers.info())
display(df_sellers.describe())
print("Soma dos valores NULL de cada coluna - ")
display(df_sellers.isnull().sum())
print("Soma dos valores NAN de cada coluna - ")
display(df_sellers.isna().sum())

print(f"Soma dos valores duplicados: {df_sellers['seller_id'].duplicated().sum()}")
```

```python
#Limpeza dos dados

df_sellers['seller_zip_code_prefix'] = df_sellers['seller_zip_code_prefix'].astype(str).str.replace(r'\.0$', '', regex=True)
df_sellers['seller_zip_code_prefix'] = df_sellers['seller_zip_code_prefix'].str.zfill(5)
df_sellers['seller_city'] = df_sellers['seller_city'].str.strip()
df_sellers['seller_city'] = df_sellers['seller_city'].str.title()
df_sellers['seller_state'] = df_sellers['seller_state'].str.strip()
df_sellers['seller_zip_code_prefix'] = df_sellers['seller_zip_code_prefix'].str.strip()
display(df_sellers)
```

```python
df_sellers.to_csv("../data/processed/sellers_processed.csv")
```
