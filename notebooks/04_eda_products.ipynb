```python
import pandas as pd
import numpy as np

#Carregando csv de produtos
df_products = pd.read_csv("../data/staging/products.csv", encoding="UTF-8")

display(df_products)
```

```python
#Análise Inicial dos dados
print("Quantidade de colunas - ")

display(df_products.shape)
display(list(df_products.columns))
columns = list(df_products.columns)

print("Descrição das tabelas - ")

display(df_products.info())
display(df_products.describe())
print("Soma dos valores NULL de cada coluna - ")
display(df_products.isnull().sum())
print("Soma dos valores NAN de cada coluna - ")
display(df_products.isna().sum())
display(df_products.loc[df_products['product_category_name'].isnull()])
display(df_products.loc[df_products['product_weight_g'].isnull()])
print(f"Soma dos valores duplicados: {df_products['product_id'].duplicated().sum()}")
```

```python
#Limpeza dos dados

df_products.drop(['product_name_lenght', 'product_description_lenght', 'product_photos_qty'], axis=1, inplace=True)
df_products.fillna({'product_category_name': 'Outra'}, inplace=True)

df_products['product_category_name'] = df_products['product_category_name'].str.strip()
df_products['product_category_name'] = df_products['product_category_name'].str.title()
display(df_products)
```

```python
df_products.to_csv("../data/processed/products_processed.csv")
```
