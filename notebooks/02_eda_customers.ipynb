```python
import pandas as pd
import numpy as np

#Carregando csv de clientes
df_customers = pd.read_csv("../data/staging/customers.csv", encoding="UTF-8")

display(df_customers)
```

```python
#Análise Inicial dos dados
print("Quantidade de colunas - ")

display(df_customers.shape)
display(list(df_customers.columns))
columns = list(df_customers.columns)

print("Descrição das tabelas - ")

display(df_customers.info())
display(df_customers.describe())
print("Soma dos valores NULL de cada coluna - ")
display(df_customers.isnull().sum())
print("Soma dos valores NAN de cada coluna - ")
display(df_customers.isna().sum())

print(f"Soma dos valores duplicados: {df_customers['customer_id'].duplicated().sum()}")
print(f"Soma dos valores duplicados: {df_customers['customer_unique_id'].duplicated().sum()}")
```

```python
#Limpeza dos dados

df_customers['customer_zip_code_prefix'] = df_customers['customer_zip_code_prefix'].astype(str).str.replace(r'\.0$', '', regex=True)
df_customers['customer_zip_code_prefix'] = df_customers['customer_zip_code_prefix'].str.zfill(5)
df_customers['customer_city'] = df_customers['customer_city'].str.strip()
df_customers['customer_city'] = df_customers['customer_city'].str.title()
df_customers['customer_state'] = df_customers['customer_state'].str.strip()
df_customers['customer_zip_code_prefix'] = df_customers['customer_zip_code_prefix'].str.strip()
display(df_customers)
```

```python
df_customers.to_csv("../data/processed/customers_processed.csv")
```
