```python
import pandas as pd
import numpy as np

#Carregando csv de pedidos
df_orders = pd.read_csv("../data/staging/orders.csv", encoding="UTF-8")

display(df_orders.sample(10))
```

```python
#Análise Inicial dos dados
print("Quantidade de colunas - ")

display(df_orders.shape)
display(list(df_orders.columns))
columns = list(df_orders.columns)

print("Descrição das tabelas - ")

display(df_orders.info())
display(df_orders.describe())
print("Soma dos valores NULL de cada coluna - ")
display(df_orders.isnull().sum())
print("Soma dos valores NAN de cada coluna - ")
display(df_orders.isna().sum())

print("Soma dos valores duplicados -")

display(df_orders.duplicated().sum())
print("Categorias da coluna status - ")
display(df_orders['order_status'].unique())
```

```python
#Limpeza dos dados
print("Colunas de datas - ")
display(df_orders)
print("Convertendo tipos das colunas - ")
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'], format="%Y-%m-%d %H:%M:%S")
df_orders['order_approved_at'] = pd.to_datetime(df_orders['order_approved_at'], format="%Y-%m-%d %H:%M:%S")
df_orders['order_delivered_carrier_date'] = pd.to_datetime(df_orders['order_delivered_carrier_date'], format="%Y-%m-%d %H:%M:%S")
df_orders['order_delivered_customer_date'] = pd.to_datetime(df_orders['order_delivered_customer_date'], format="%Y-%m-%d %H:%M:%S")
df_orders['order_estimated_delivery_date'] = pd.to_datetime(df_orders['order_estimated_delivery_date'], format="%Y-%m-%d")
df_orders['order_status'] = df_orders['order_status'].str.strip()
df_orders['order_status'] = df_orders['order_status'].str.title()
order_status_categories = list(df_orders['order_status'].unique())
display(df_orders.info())
```

```python
print("Analisando a coluna order_approved_at - ")
print("Status dos pedidos com data de aprovação nula - ")

for status in order_status_categories:
    quantity = df_orders.loc[(df_orders['order_approved_at'].isnull()) & (df_orders['order_status'] == status)].shape[0]
    if(quantity > 0):
        print(status)
        display(quantity)

#Provavelmente faz parte da regra de negócio deixar colunas de datas nulas quando o status referente permite
```
